## 과제 내용
은닉층이 1개인 모델과 2개인 모델을 각각 작성하고 구조 차이를 설명합니다.

## 제출 방법
모든 셀을 실행해 출력이 보이는 상태로 저장한 뒤 `week06_학번_이름.ipynb` 로 제출합니다.


```python
import torch
import torch.nn as nn
import pandas as pd

torch.manual_seed(0)
```

## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
# 은닉층 1개 모델 정의
class OneHidden(nn.Module):
    # 모델의 층 생성
    def __init__(self):
        super().__init__()             # 부모 클래스 초기화
        self.fc1 = nn.Linear(5, 16)    # 입력층 → 은닉층
        self.relu = nn.ReLU()          # 활성화 함수 생성
        self.fc2 = nn.Linear(16, 3)    # 은닉층 → 출력층

    # 순전파 정의
    def forward(self, x):
        x = self.fc1(x)                # 첫 번째 Linear 연산
        x = self.relu(x)               # ReLU 활성화 함수 적용
        x = self.fc2(x)                # 두 번째 Linear 연산
        return x                       # 최종 출력 반환

# 모델 생성 및 구조 출력
model1 = OneHidden()
print(model1)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
# 답안
# 은닉층 2개 모델 정의
class TwoHidden(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)     # 입력층 → 첫 번째 은닉층
        self.relu1 = nn.ReLU()          # 첫 번째 활성화 함수
        self.fc2 = nn.Linear(16, 16)    # 첫 번째 → 두 번째 은닉층
        self.relu2 = nn.ReLU()          # 두 번째 활성화 함수
        self.fc3 = nn.Linear(16, 3)     # 두 번째 은닉층 → 출력층

    def forward(self, x):
        x = self.relu1(self.fc1(x))     # 첫 번째 은닉층 연산
        x = self.relu2(self.fc2(x))     # 두 번째 은닉층 연산
        return self.fc3(x)              # 최종 출력 반환

# 모델 생성 및 구조 출력
model2 = TwoHidden()
print(model2)
```

## 문제 3. 두 모델의 구조 비교표 (30점)
아래 항목으로 표를 만들어 출력하세요.

| 항목 | 은닉층 1개 | 은닉층 2개 |
| --- | --- | --- |
| Linear 층 수 | | |
| 전체 파라미터 수 | | |
| 활성화 함수 개수 | | |
| 출력 shape | | |


```python
# 답안 — pd.DataFrame 으로 만들어 출력하세요
# 임의 입력 데이터 생성
x = torch.randn(8, 5)

# 두 모델의 구조 비교
table = {}

for name, model in [("은닉층 1개", model1), ("은닉층 2개", model2)]:
    table[name] = [
        sum(isinstance(m, nn.Linear) for m in model.modules()),
        sum(p.numel() for p in model.parameters()),
        sum(isinstance(m, nn.ReLU) for m in model.modules()),
        tuple(model(x).shape)
    ]

# 비교표 출력
pd.DataFrame(table, index=[
    "Linear 층 수",
    "전체 파라미터 수",
    "활성화 함수 개수",
    "출력 shape"
])
```

## 문제 4. 구조 차이 설명 (20점)
아래 세 가지를 각각 두세 줄로 설명하세요.

1. 층을 늘리면 파라미터 수가 왜 늘어나는가
2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유
3. 층을 늘리는 것이 항상 좋은가 (근거를 함께)


**답:**

1.각 Linear 층에는 가중치와 편향이 존재하므로 층을 추가하면 학습해야 할 파라미터도 늘어난다. 실제로 은닉층을 1개에서 2개로 늘리자 파라미터 수가 147개에서 419개로 증가하였다.

2.활성화 함수가 없으면 여러 Linear 층을 쌓아도 하나의 선형 변환으로 합칠 수 있다. 따라서 층을 늘려도 복잡한 비선형 관계를 학습하기 어렵다.

3.층을 늘리면 복잡한 패턴을 학습할 수 있지만, 파라미터 수와 연산량도 증가한다. 또한 과적합이 발생할 수 있으므로 데이터와 문제의 복잡도에 맞게 층 수를 결정해야 한다.
